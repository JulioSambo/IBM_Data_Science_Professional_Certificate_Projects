An example of folium.Circle:


`folium.Circle(coordinate, radius=1000, color='#000000', fill=True).add_child(folium.Popup(...))`


An example of folium.Marker:


`folium.map.Marker(coordinate, icon=DivIcon(icon_size=(20,20),icon_anchor=(0,0), html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % 'label', ))`


```python
# Initial the map
site_map = folium.Map(location=nasa_coordinate, zoom_start=10)

# Loop through each row in launch_sites_df to add circles and popups
for index, row in launch_sites_df.iterrows():
    # 1. Grab coordinates and site name
    coordinate = [row['Lat'], row['Long']]
    site_name = row['Launch Site']
    
    # 2. Add a circle marker with a popup label for the site
    circle = folium.Circle(
        location=coordinate, 
        radius=1000, 
        color='#d35400', 
        fill=True
    ).add_child(folium.Popup(site_name))
    
    # 3. Add a text label directly on the map (fixed font-size unit here)
    marker = folium.map.Marker(
        location=coordinate,
        icon=DivIcon(
            icon_size=(20, 20),
            icon_anchor=(0, 0),
            html=f'<div style="font-size: 12pt; color:#d35400;"><b>{site_name}</b></div>',
        )
    )
    
    # Attach both elements to the map
    site_map.add_child(circle)
    site_map.add_child(marker)

# Display the final map
site_map
```

Next, let's create markers for all launch records.
If a launch was successful `(class=1)`, then we use a green marker and if a launch was failed, we use a red marker `(class=0)`


Note that a launch only happens in one of the four launch sites, which means many launch records will have the exact same coordinate. Marker clusters can be a good way to simplify a map containing many markers having the same coordinate.


Let's first create a `MarkerCluster` object


```python
marker_cluster = MarkerCluster()
```

*TODO:* Create a new column in `spacex_df` dataframe called `marker_color` to store the marker colors based on the `class` value


```python

# Apply a function to check the value of `class` column
def check_class(Class):
    if Class ==1:
        return 'green'
    else:
        return 'red'
spacex_df['marker_color'] = spacex_df['class'].apply(check_class)
spacex_df.tail(5)


# spacex_df['marker_color'] = spacex_df['class'].apply(lambda x: 'green' if x == 1 else 'red')
# spacex_df.tail(5)
```

*TODO:* For each launch result in `spacex_df` data frame, add a `folium.Marker` to `marker_cluster`


```python
# Add marker_cluster to current site_map
site_map.add_child(marker_cluster)

# for each row in spacex_df data frame
# create a Marker object with its coordinate
# and customize the Marker's icon property to indicate if this launch was successed or failed
for index, record in spacex_df.iterrows():
    # 1. Extract the latitude and longitude coordinate pair
    coordinate = [record['Lat'], record['Long']]
    
    # 2. Create the Marker object with a customized icon color based on success/failure
    marker = folium.Marker(
        location=coordinate,
        icon=folium.Icon(color='white', icon_color=record['marker_color']),
        popup=f"Site: {record['Launch Site']}<br>Status: {'Success' if record['class'] == 1 else 'Fail'}"
    )
    
    # 3. Add the marker to the cluster group
    marker_cluster.add_child(marker)

# Display the map with your newly populated cluster
site_map
```

Next, we need to explore and analyze the proximities of launch sites.


Let's first add a `MousePosition` on the map to get coordinate for a mouse over a point on the map. As such, while you are exploring the map, you can easily find the coordinates of any points of interests (such as railway)


```python
# Add Mouse Position to get the coordinate (Lat, Long) for a mouse over on the map
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)
site_map
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

*TODO:* Mark down a point on the closest coastline using MousePosition and calculate the distance between the coastline point and the launch site.


```python
# find coordinate of the closet coastline
# e.g.,: Lat: 28.56367  Lon: -80.57163
# distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)

# Define coordinates
launch_site_lat = 28.563197
launch_site_lon = -80.576820
coastline_lat = 28.56367
coastline_lon = -80.57163

# Calculate distance
distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
print(f"Distance to coastline: {distance_coastline:.2f} KM")
```

```python
# Create and add a folium.Marker on your selected closest coastline point on the map
# Display the distance between coastline point and launch site using the icon property 
# for example
# distance_marker = folium.Marker(
#    coordinate,
#    icon=DivIcon(
#        icon_size=(20,20),
#        icon_anchor=(0,0),
#        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance),
#        )
#    )
from folium.features import DivIcon

# Coordinates pair for the marker location
coastline_coordinate = [coastline_lat, coastline_lon]

# Create the distance marker
distance_marker = folium.Marker(
    coastline_coordinate,
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html=f'<div style="font-size: 12pt; color:#d35400;"><b>{distance_coastline:.2f} KM</b></div>',
    )
)

# Add it to the map
site_map.add_child(distance_marker)
```

```python
# Create a marker with distance to a closest city, railway, highway, etc.
# Draw a line between the marker to the launch site
```

```python
# 1. Define your base Launch Site coordinates (CCAFS SLC-40)
launch_site_lat = 28.563197
launch_site_lon = -80.576820

# 2. Define coordinates for the closest points of interest (POIs)
# (You can use these exact coordinates found via MousePosition)
poi_coordinates = {
    'Highway (Samuel C Phillips Pkwy)': [28.56385, -80.57085],
    'Railway (NASA Railroad)': [28.57205, -80.58525],
    'City (Titusville)': [28.6122, -80.8076]
}

# 3. Loop through each POI to calculate distance, add a marker, and draw a line
for poi_name, poi_coord in poi_coordinates.items():
    # Calculate the distance using the lab's built-in function
    distance = calculate_distance(launch_site_lat, launch_site_lon, poi_coord[0], poi_coord[1])
    
    # Create and add a text marker on the map to display the distance
    distance_marker = folium.Marker(
        poi_coord,
        icon=folium.features.DivIcon(
            icon_size=(20,20),
            icon_anchor=(0,0),
            html=f'<div style="font-size: 12pt; color:#d35400;"><b>{distance:.2f} KM</b></div>',
        )
    )
    site_map.add_child(distance_marker)
    
    # Create and add a line connecting the launch site to the POI
    line_coordinates = [[launch_site_lat, launch_site_lon], poi_coord]
    lines = folium.PolyLine(locations=line_coordinates, weight=1, color='green')
    site_map.add_child(lines)

# Display the map updated with lines to the Highway, Railway, and City
site_map
```
